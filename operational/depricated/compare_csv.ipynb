```python
import pandas as pd

df_1 = pd.read_csv("animal_sessions_rawData.csv")
df_2 = pd.read_csv("animal_sessions_rawDataEdited.csv")
```

```python
merged = pd.merge(df_1, df_2, on=["id", "date"], how="outer", suffixes=("_in_1", "_in_2"))
df_3 = merged[merged["n_trials_in_1"] != merged["n_trials_in_2"]][["id", "date", "n_trials_in_1", "n_trials_in_2"]]
```

```python
# Order rows by date
df_3 = df_3.sort_values(by="date")
```

```python
# Save df as csv
df_3.to_csv("discrepancies.csv", index=False)
```
